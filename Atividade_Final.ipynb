# **Atividade Final – Web Scraping e Coleta de Dados Automatizada**

##  **Objetivo da Atividade**
Nesta atividade, o aluno deverá aplicar técnicas de Web Scraping e automação de navegação para coletar dados reais de marketplaces, comparar preços e gerar uma análise final.  
O objetivo é simular uma situação real: **fazer uma pesquisa de preços completa para a compra de materiais de reforma doméstica**, utilizando ferramentas modernas de raspagem de dados.

---

## **Contexto**
Você recebeu uma lista com **25 itens de materiais** necessários para realizar uma reforma completa em uma residência.  
Seu objetivo é descobrir **onde seria mais barato** comprar todos os itens.

Para isso, você deve construir **uma automação capaz de pesquisar cada item em 3 marketplaces diferentes**, coletar dados, comparar e gerar uma análise final em PDF.  
Ao final, será necessário **enviar o relatório por e-mail automaticamente**, além de registrar logs e gerar um vídeo da execução.

---

## **Ferramentas Permitidas**
Você pode usar **qualquer ferramenta de automação**, incluindo:

- Selenium  
- SeleniumBase  
- Playwright  
- Scrapy  
- Requests + BeautifulSoup

---

##  **Tarefas Obrigatórias**

### 1 **Receber a lista de 25 itens**

Cada item pode estar no formato:  
*“20 Sacos de argamassa de 15kg”*  
ou  
*“50 metros de cabo elétrico 2,5mm (cor verde)”*

---

### 2 **Pesquisar cada item nos 3 marketplaces**

---

### 3 **Coletar informações**
Para **cada site**, capturar:

- **Nome do produto**
- **Preço**
- **Quantidade**, caso venha separada do nome  
- **URL** (opcional, mas recomendado)
- **Página de onde o item foi coletado**

Regras:

- Coletar **os 3 primeiros produtos exibidos** na busca. 
- Coletar o produto  mais barato da pesquisa entre as  **as três primeiras**. (Caso a pesquisa gere menos de três paginas, pegue as duas primeiras. Se existir apenas uma página, use ela). 
- Lembrando que os produtos coletados devem se aproximar o maximo possível do nome presente na lista.

---

### 4 **Estruturar os dados**
Monte uma estrutura organizada, como um:

- **Pandas DataFrame**  
ou
- **Lista de dicionários**  

Organize de forma que seja possível comparar:

| item | site | nome | preço | quantidade | link |

---

### 5 **Análise descritiva**
O aluno deve:

- Comparar o preço médio por item entre sites  
- Identificar o site mais barato para cada item  
- Identificar o **site mais vantajoso para toda a compra**  
- Criar estatísticas como:
  - menor preço por item  
  - maior preço por item  
  - variação percentual  
  - economia total escolhendo sempre o menor preço  

---

### 6 **Gerar um PDF da análise**
O PDF deve incluir:

- Gráficos  
- Tabelas  
- Conclusões  
- Recomendação final de onde comprar  

Ferramentas possíveis:

- matplotlib + PdfPages
- reportlab  
- Pandas + DataFrame.to_latex + compilação  

---

### 7 **Enviar o PDF por e-mail**
O aluno deve enviar automaticamente:

- O PDF completo  
- Um **resumo textual** com os principais achados

Pode utilizar:

- `smtplib`
- SeleniumBase

Dica: gerar senha para o gmail: https://myaccount.google.com/apppasswords
---

### 8 **Gerar logs**
O sistema deve registrar:

- A data/hora da execução  
- Cada item pesquisado  
- Erros encontrados  
- Sites visitados  
- Status final  

Formato sugerido: arquivo `.log` ou `.txt`.

---

### 9 **Gravar a tela da automação**
O aluno deve entregar um **vídeo mostrando a automação** funcionando do início ao fim.

---

### 10 **Entrega Final**
O aluno deve entregar um arquivo **ZIP** contendo os arquivos e o vídeo.
